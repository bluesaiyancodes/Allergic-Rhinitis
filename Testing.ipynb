```python
import cv2
import pandas as pd
import numpy as np
```

```python
dataF = pd.read_csv('ARStats.csv')
```

```python
dataF.tail(21)
```

```python
data = dataF.to_numpy()
```

```python
data20 = data[len(data)-20:]
```

```python
len(data[:])
```

```python
accInc = []
accXc = []
for i in range(len(data20)):
    if data20[i][2] == 'inception':
        accInc.append(data20[i][-1])
    elif data[i][2] == "xception":
        accXc.append(data20[i][-1])
```

```python
print("%.2f%% (+/- %.2f%%)" % (np.mean(accInc), np.std(accInc)))  
```

```python
print("%.2f%% (+/- %.2f%%)" % (np.mean(accXc), np.std(accXc)))  
```
